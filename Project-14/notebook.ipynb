# Project 14 - OPEN-ARC

In this project, we’ll be classifying images of human faces into two categories: real or fake. The aim is to determine with accuracy whether a face is genuine or computer-generated.

```python
import os
import random
import copy
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.amp import autocast, GradScaler
from torchvision import datasets, models
from torchvision.models import ResNet18_Weights
from torchvision.transforms.functional import resize as tv_resize
from PIL import Image
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from tqdm import tqdm

torch.backends.cudnn.benchmark = True

# ---------------------------------------------------------
# 1. Configuration
# ---------------------------------------------------------
IMG_SIZE = 224              # matches ImageNet pretraining input size — preserves fine artifact detail
BATCH_SIZE = 64

EPOCHS_HEAD = 4              # phase 1: train only the new classifier head, backbone frozen
LR_HEAD = 1e-3

EPOCHS_FINETUNE = 20          # phase 2: unfreeze everything, fine-tune end-to-end
LR_FINETUNE = 2e-4
FT_WARMUP_EPOCHS = 2
MIN_LR = 1e-6
EARLY_STOP_PATIENCE = 6       # stop phase 2 early if val_acc stalls

CROP_PAD = 10
BRIGHTNESS_JITTER = 0.10
CONTRAST_JITTER = 0.10
LABEL_SMOOTHING = 0.05
GRAD_CLIP_NORM = 1.0
N_SAMPLES_GRID = 16

BASE_DIR = '/kaggle/input/datasets/sachchitkunichetty/rvf10k/rvf10k'
TRAIN_DIR = os.path.join(BASE_DIR, 'train')
VALID_DIR = os.path.join(BASE_DIR, 'valid')

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Primary Device: {device} | GPU count: {torch.cuda.device_count()}")

# ---------------------------------------------------------
# 2. Load dataset into memory as uint8, move to GPU once
#    (uint8 storage = 4x smaller than float32, still fits at 224px easily)
# ---------------------------------------------------------
MEAN = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1)
STD = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1)

def load_split_to_tensor(root_dir, img_size):
    ds = datasets.ImageFolder(root_dir)
    n = len(ds.samples)
    imgs = torch.empty((n, 3, img_size, img_size), dtype=torch.uint8)
    labels = torch.empty((n,), dtype=torch.long)

    for i, (path, label) in enumerate(tqdm(ds.samples, desc=f"Loading {os.path.basename(root_dir)}")):
        img = Image.open(path).convert("RGB")
        img = tv_resize(img, [img_size, img_size])
        arr = torch.from_numpy(np.array(img)).permute(2, 0, 1).to(torch.uint8)
        imgs[i] = arr
        labels[i] = label

    return imgs, labels, ds.classes

train_imgs_cpu, train_labels_cpu, class_names = load_split_to_tensor(TRAIN_DIR, IMG_SIZE)
valid_imgs_cpu, valid_labels_cpu, _ = load_split_to_tensor(VALID_DIR, IMG_SIZE)

print(f"Training Images: {len(train_imgs_cpu)} | Validation Images: {len(valid_imgs_cpu)}")
print(f"Classes: {class_names}")
print(f"Train tensor size (uint8, GPU-resident): {train_imgs_cpu.element_size() * train_imgs_cpu.nelement() / 1e9:.2f} GB")

train_imgs = train_imgs_cpu.to(device, non_blocking=True)   # uint8, small footprint
train_labels = train_labels_cpu.to(device, non_blocking=True)
valid_imgs = valid_imgs_cpu.to(device, non_blocking=True)
valid_labels = valid_labels_cpu.to(device, non_blocking=True)
mean_gpu = MEAN.to(device)
std_gpu = STD.to(device)
del train_imgs_cpu, valid_imgs_cpu

n_train = train_imgs.size(0)
n_valid = valid_imgs.size(0)

# ---------------------------------------------------------
# 3. GPU-side augmentation + normalization (operates on uint8 input)
# ---------------------------------------------------------
def augment_batch_gpu(batch_uint8):
    batch = batch_uint8.float() / 255.0
    b, c, h, w = batch.shape

    flip_mask = torch.rand(b, device=batch.device) < 0.5
    batch[flip_mask] = batch[flip_mask].flip(-1)

    padded = F.pad(batch, (CROP_PAD, CROP_PAD, CROP_PAD, CROP_PAD), mode='reflect')
    max_off = 2 * CROP_PAD
    off_x = random.randint(0, max_off)
    off_y = random.randint(0, max_off)
    batch = padded[:, :, off_y:off_y + h, off_x:off_x + w]

    brightness = 1.0 + (torch.rand(b, 1, 1, 1, device=batch.device) - 0.5) * (2 * BRIGHTNESS_JITTER)
    contrast = 1.0 + (torch.rand(b, 1, 1, 1, device=batch.device) - 0.5) * (2 * CONTRAST_JITTER)
    batch = batch * brightness
    mean_per_img = batch.mean(dim=(1, 2, 3), keepdim=True)
    batch = (batch - mean_per_img) * contrast + mean_per_img
    batch = batch.clamp(0, 1)

    batch = (batch - mean_gpu) / std_gpu
    return batch

def normalize_only_gpu(batch_uint8):
    batch = batch_uint8.float() / 255.0
    return (batch - mean_gpu) / std_gpu

# ---------------------------------------------------------
# 4. Model: pretrained ResNet18 backbone + new classifier head
# ---------------------------------------------------------
try:
    weights = ResNet18_Weights.IMAGENET1K_V1
    model = models.resnet18(weights=weights)
    print("Loaded ImageNet-pretrained ResNet18 weights.")
except Exception as e:
    print(f"Could not download pretrained weights ({e}). "
          "Falling back to random init — check that 'Internet' is turned ON "
          "in the Kaggle notebook's Settings panel for the accuracy boost to apply.")
    model = models.resnet18(weights=None)

in_features = model.fc.in_features
model.fc = nn.Sequential(
    nn.Dropout(0.3),
    nn.Linear(in_features, 2)
)
model.to(device)
print("Model device:", next(model.parameters()).device)

def set_backbone_trainable(model, trainable):
    for name, module in model.named_children():
        if name != 'fc':
            for p in module.parameters():
                p.requires_grad = trainable

criterion = nn.CrossEntropyLoss(label_smoothing=LABEL_SMOOTHING)
scaler = GradScaler('cuda')

def set_lr(optimizer, lr):
    for g in optimizer.param_groups:
        g['lr'] = lr

@torch.no_grad()
def evaluate_quick():
    model.eval()
    correct, total = 0, 0
    for start in range(0, n_valid, BATCH_SIZE):
        images = normalize_only_gpu(valid_imgs[start:start + BATCH_SIZE])
        labels = valid_labels[start:start + BATCH_SIZE]
        outputs = model(images)
        preds = torch.argmax(outputs, dim=1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)
    return correct / total

def train_one_epoch(optimizer, lr, epoch_idx, total_epochs, phase_label, freeze_bn=False):
    model.train()
    if freeze_bn:
        # keep BatchNorm running stats frozen while backbone weights are frozen
        for name, module in model.named_children():
            if name != 'fc':
                module.eval()
    set_lr(optimizer, lr)

    running_loss = 0.0
    corrects = 0
    total = 0
    perm = torch.randperm(n_train, device=device)
    loop = tqdm(range(0, n_train, BATCH_SIZE),
                desc=f"{phase_label} [{epoch_idx+1}/{total_epochs}] lr={lr:.2e}", leave=True)

    for start in loop:
        idx = perm[start:start + BATCH_SIZE]
        images = augment_batch_gpu(train_imgs[idx])
        labels = train_labels[idx]

        optimizer.zero_grad()
        with autocast('cuda'):
            outputs = model(images)
            loss = criterion(outputs, labels)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP_NORM)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item() * images.size(0)
        _, preds = torch.max(outputs, 1)
        corrects += torch.sum(preds == labels.data)
        total += labels.size(0)
        loop.set_postfix(loss=running_loss / total, acc=corrects.double().item() / total)

    return corrects.double().item() / total, running_loss / total

# ---------------------------------------------------------
# 5. Phase 1 — train classifier head only, backbone frozen
# ---------------------------------------------------------
best_val_acc = 0.0
best_state = None
history = {"train_acc": [], "val_acc": [], "train_loss": [], "lr": []}

set_backbone_trainable(model, False)
head_optimizer = optim.AdamW(
    [p for p in model.parameters() if p.requires_grad], lr=LR_HEAD, weight_decay=1e-4
)

for epoch in range(EPOCHS_HEAD):
    train_acc, train_loss = train_one_epoch(
        head_optimizer, LR_HEAD, epoch, EPOCHS_HEAD, "Head", freeze_bn=True
    )
    val_acc = evaluate_quick()
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)
    history["train_loss"].append(train_loss)
    history["lr"].append(LR_HEAD)
    print(f"  -> val_acc={val_acc:.4f} (best={best_val_acc:.4f})")
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_state = copy.deepcopy(model.state_dict())

# ---------------------------------------------------------
# 6. Phase 2 — unfreeze everything, fine-tune end-to-end
# ---------------------------------------------------------
set_backbone_trainable(model, True)
ft_optimizer = optim.AdamW(model.parameters(), lr=LR_FINETUNE, weight_decay=1e-4)

def lr_at_ft_epoch(epoch):
    if epoch < FT_WARMUP_EPOCHS:
        return LR_FINETUNE * (epoch + 1) / FT_WARMUP_EPOCHS
    progress = (epoch - FT_WARMUP_EPOCHS) / max(1, EPOCHS_FINETUNE - FT_WARMUP_EPOCHS)
    cosine = 0.5 * (1 + np.cos(np.pi * progress))
    return MIN_LR + (LR_FINETUNE - MIN_LR) * cosine

epochs_since_improvement = 0
for epoch in range(EPOCHS_FINETUNE):
    lr = lr_at_ft_epoch(epoch)
    train_acc, train_loss = train_one_epoch(
        ft_optimizer, lr, epoch, EPOCHS_FINETUNE, "Finetune", freeze_bn=False
    )
    val_acc = evaluate_quick()
    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)
    history["train_loss"].append(train_loss)
    history["lr"].append(lr)
    print(f"  -> val_acc={val_acc:.4f} (best={best_val_acc:.4f})")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_state = copy.deepcopy(model.state_dict())
        epochs_since_improvement = 0
    else:
        epochs_since_improvement += 1
        if epochs_since_improvement >= EARLY_STOP_PATIENCE:
            print(f"No improvement for {EARLY_STOP_PATIENCE} epochs — stopping early.")
            break

if best_state is not None:
    model.load_state_dict(best_state)
    print(f"\nLoaded best checkpoint with val_acc={best_val_acc:.4f}")

# ---------------------------------------------------------
# 7. Training curves
# ---------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(history["train_acc"], label="train")
axes[0].plot(history["val_acc"], label="val")
axes[0].axvline(EPOCHS_HEAD - 0.5, color='gray', linestyle='--', label="unfreeze")
axes[0].set_title("Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].legend()
axes[1].plot(history["train_loss"], label="train loss")
axes[1].axvline(EPOCHS_HEAD - 0.5, color='gray', linestyle='--')
axes[1].set_title("Loss")
axes[1].set_xlabel("Epoch")
axes[1].legend()
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 8. Final Evaluation (best checkpoint)
# ---------------------------------------------------------
model.eval()
all_preds, all_labels = [], []

with torch.no_grad():
    for start in tqdm(range(0, n_valid, BATCH_SIZE), desc="Validating"):
        images = normalize_only_gpu(valid_imgs[start:start + BATCH_SIZE])
        labels = valid_labels[start:start + BATCH_SIZE]
        outputs = model(images)
        _, preds = torch.max(outputs, 1)
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

acc = accuracy_score(all_labels, all_preds)
prec = precision_score(all_labels, all_preds)
rec = recall_score(all_labels, all_preds)
f1 = f1_score(all_labels, all_preds)

print("\n--- Validation Metrics (best checkpoint) ---")
print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall:    {rec:.4f}")
print(f"F1 Score:  {f1:.4f}")

cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(6, 4))
plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
plt.title("Confusion Matrix")
plt.colorbar()
tick_marks = np.arange(2)
plt.xticks(tick_marks, class_names)
plt.yticks(tick_marks, class_names)
for i in range(2):
    for j in range(2):
        plt.text(j, i, str(cm[i, j]), ha="center", va="center", color="red")
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 9. Inference Grid
# ---------------------------------------------------------
indices = random.sample(range(n_valid), N_SAMPLES_GRID)
fig, axes = plt.subplots(4, 4, figsize=(15, 15))
axes = axes.flatten()

with torch.no_grad():
    for i, idx in enumerate(indices):
        img_uint8 = valid_imgs[idx].unsqueeze(0)
        label = valid_labels[idx].item()
        img_norm = normalize_only_gpu(img_uint8)

        output = model(img_norm)
        pred = torch.argmax(output, dim=1).item()

        img_np = img_uint8.squeeze(0).cpu().numpy().transpose((1, 2, 0))  # already 0-255 uint8

        axes[i].imshow(img_np)
        true_label = class_names[label]
        pred_label = class_names[pred]
        color = "green" if true_label == pred_label else "red"
        axes[i].set_title(f"True: {true_label}\nPred: {pred_label}", color=color, fontweight='bold')
        axes[i].axis('off')

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# 10. Model Export (best checkpoint)
# ---------------------------------------------------------
save_path = 'novel_fast_face_detector_resnet18.pth'
torch.save(model.state_dict(), save_path)
print(f"\nModel exported successfully to: {save_path} (val_acc={best_val_acc:.4f})")
```
