```
# Step 2: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# Set plot style and figure size defaults
plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = (8, 5)
print("Libraries imported successfully!")
```

```
# Step 3: Load dataset from Hugging Face
dataset = load_dataset("dair-ai/emotion", split="train[:1000]")
df = pd.DataFrame(dataset)

# Mapping emotion numeric labels to human-readable emotion names
label_map = {0: 'sadness', 1: 'joy', 2: 'love', 3: 'anger', 4: 'fear', 5: 'surprise'}
df['emotion_name'] = df['label'].map(label_map)

print("Dataset shape:", df.shape)
df.head()
```

```
# Step 4: Compute EDA Features
# Compute character length and word count per example
df['char_length'] = df['text'].apply(len)
df['word_count'] = df['text'].apply(lambda x: len(x.split()))

# Summary statistics for numerical columns
df[['char_length', 'word_count']].describe()
```

```
# Step 5: Visualization 1 - Class Distribution
# Count frequency of each emotion label
label_counts = df['emotion_name'].value_counts()

plt.figure(figsize=(8, 4))
plt.bar(label_counts.index, label_counts.values, color='skyblue', edgecolor='black')
plt.title("Distribution of Emotion Labels in Dataset", fontsize=14, fontweight='bold')
plt.xlabel("Emotion Class", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```
# Step 6: Visualization 2 - Word Count Histogram
plt.figure(figsize=(8, 4))
plt.hist(df['word_count'], bins=25, color='teal', edgecolor='black', alpha=0.7)
plt.title("Histogram of Word Counts per Entry", fontsize=14, fontweight='bold')
plt.xlabel("Word Count", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.axvline(df['word_count'].mean(), color='red', linestyle='dashed', linewidth=2, label=f"Mean ({df['word_count'].mean():.1f})")
plt.legend()
plt.tight_layout()
plt.show()
```

```
# Step 7: Visualization 3 - Word Count Box Plot by Emotion Class
emotions = list(df['emotion_name'].unique())
word_counts_by_emotion = [df[df['emotion_name'] == emo]['word_count'] for emo in emotions]

plt.figure(figsize=(9, 5))
plt.boxplot(word_counts_by_emotion, tick_labels=emotions, patch_artist=True)
plt.title("Word Count Distribution Across Emotion Categories", fontsize=14, fontweight='bold')
plt.xlabel("Emotion Class", fontsize=12)
plt.ylabel("Word Count", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
